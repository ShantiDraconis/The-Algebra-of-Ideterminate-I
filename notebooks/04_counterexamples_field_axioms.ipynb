# Counterexamples to Field Axioms

This notebook provides computational verification of counterexamples showing where classical field axioms fail in The Algebra of Indeterminate I.

## Setup

```python
import json
import os
from typing import Dict, List, Any

# Define element representation
class Element:
    def __init__(self, name, value=None):
        self.name = name
        self.value = value if value is not None else name
    
    def __repr__(self):
        return self.name
    
    def __eq__(self, other):
        return isinstance(other, Element) and self.name == other.name

# Define elements
ZERO = Element('0', 0)
ONE = Element('1', 1)
I = Element('I')
NEG_I = Element('-I')

print("✓ Elements defined")
```

## Operations

```python
def add(a: Element, b: Element) -> Element:
    """Addition in the algebra"""
    if a == ZERO:
        return b
    if b == ZERO:
        return a
    if a == I and b == I:
        return I
    if a == I and b == NEG_I:
        return ZERO
    if a == NEG_I and b == I:
        return ZERO
    return Element(f"({a}+{b})")

def multiply(a: Element, b: Element) -> Element:
    """Multiplication in the algebra"""
    if a == ZERO or b == ZERO:
        return ZERO
    if a == ONE:
        return b
    if b == ONE:
        return a
    if a == I and b == I:
        return I
    return Element(f"({a}·{b})")

print("✓ Operations defined")
```

## Counterexample Collection

We'll collect counterexamples in a structured format for export.

```python
counterexamples = []

def add_counterexample(id: str, name: str, axiom: str, 
                       description: str, proof_steps: List[str],
                       conclusion: str):
    """Add a counterexample to the collection"""
    counterexamples.append({
        'id': id,
        'name': name,
        'axiom_violated': axiom,
        'description': description,
        'proof_steps': proof_steps,
        'conclusion': conclusion
    })
```

## CE1: Multiplicative Inverse Fails for I

```python
print("=== CE1: No Multiplicative Inverse for I ===")
print("\nField Axiom: ∀a ≠ 0: ∃a⁻¹: a · a⁻¹ = 1")
print("\nProof by contradiction:")

steps = []

# Step 1
print("\n1. Assume I⁻¹ exists such that I · I⁻¹ = 1")
steps.append("Assume I⁻¹ exists such that I · I⁻¹ = 1")

# Step 2
print("2. By P8 (multiplicative identity): I = I · 1")
steps.append("By P8: I = I · 1")

# Step 3
print("3. Substitute I · I⁻¹ for 1: I = I · (I · I⁻¹)")
steps.append("Substitute: I = I · (I · I⁻¹)")

# Step 4
print("4. By P7 (associativity): I = (I · I) · I⁻¹")
steps.append("By P7: I = (I · I) · I⁻¹")

# Step 5
print("5. By P11 (I² = I): I = I · I⁻¹")
result = multiply(I, I)
print(f"   Verification: I · I = {result}")
assert result == I, "P11 violated!"
steps.append("By P11 (I² = I): I = I · I⁻¹")

# Step 6
print("6. By assumption: I = 1")
steps.append("By assumption: I = 1")

# Step 7
print("7. But this contradicts P10 (I ≠ 1)")
print("\n∴ I⁻¹ does not exist. ✓")
steps.append("But this contradicts P10 (I ≠ 1)")

add_counterexample(
    'CE1',
    'No Multiplicative Inverse for I',
    'Field Axiom: All non-zero elements have multiplicative inverse',
    'I has no multiplicative inverse',
    steps,
    'Therefore I⁻¹ does not exist'
)

print("\n" + "="*50)
```

## CE2: Cancellation Law Fails

```python
print("=== CE2: Cancellation Law Fails ===")
print("\nField Axiom: ∀a,b,c: (c ≠ 0 ∧ c·a = c·b) ⟹ a = b")
print("\nCounterexample:")

steps = []

# Setup
print("\nLet c = I, a = I, b = 1")
c, a, b = I, I, ONE
steps.append("Let c = I, a = I, b = 1")

# Computation
ca = multiply(c, a)
cb = multiply(c, b)

print(f"\nc · a = I · I = {ca}")
print(f"c · b = I · 1 = {cb}")
steps.append(f"I · I = {ca}")
steps.append(f"I · 1 = {cb}")

# Verify equality
print(f"\nTherefore: c·a = c·b ({ca} = {cb})")
assert ca == cb, "Products not equal!"
steps.append(f"Therefore c·a = c·b")

# Show contradiction
print(f"But: a ≠ b (I ≠ 1 by P10)")
assert a != b, "Elements should be different!"
steps.append("But a ≠ b (I ≠ 1 by P10)")

print("\n∴ Cancellation law does not hold. ✓")

add_counterexample(
    'CE2',
    'Cancellation Law Fails',
    'Field Axiom: Cancellation law for multiplication',
    'Cannot cancel I from both sides of equation',
    steps,
    'Therefore cancellation law fails for I'
)

print("\n" + "="*50)
```

## CE4: Additive Idempotency Breaks Standard Group Reasoning

```python
print("=== CE4: Additive Idempotency ===")
print("\nGroup Property: ∀a,b: a + a = a + b ⟹ a = b")
print("\nCounterexample:")

steps = []

# Computation
i_plus_i = add(I, I)
i_plus_0 = add(I, ZERO)

print(f"\nI + I = {i_plus_i}")
print(f"I + 0 = {i_plus_0}")
steps.append(f"I + I = {i_plus_i} (by P12)")
steps.append(f"I + 0 = {i_plus_0} (by P4)")

assert i_plus_i == i_plus_0, "Results not equal!"
print(f"\nTherefore: I + I = I + 0")
steps.append("Therefore I + I = I + 0")

print(f"But: I ≠ 0 (by P10)")
assert I != ZERO, "I should not equal 0!"
steps.append("But I ≠ 0 (by P10)")

print("\nIn a normal group, we could add (-I) to both sides,")
print("which would lead to I = 0. But this contradicts P10.")
steps.append("Normal group reasoning fails due to idempotency")

print("\n∴ Standard group cancellation reasoning fails. ✓")

add_counterexample(
    'CE4',
    'Additive Idempotency Breaks Group Reasoning',
    'Group Axiom: Cancellation in groups',
    'I + I = I + 0 but I ≠ 0',
    steps,
    'Standard group cancellation fails due to idempotency of I'
)

print("\n" + "="*50)
```

## CE7: Polynomial Equations Have More Roots Than Expected

```python
print("=== CE7: Polynomial Roots ===")
print("\nField Property: Polynomial x² - x = 0 has at most 2 roots")
print("\nCounterexample - equation x² = x:")

steps = []

# Find roots
roots = []

print("\nTesting x = 0:")
x = ZERO
x_squared = multiply(x, x)
print(f"  0² = {x_squared}")
print(f"  0² = 0? {x_squared == x}")
if x_squared == x:
    roots.append('0')
    steps.append("x = 0 is a root: 0² = 0")

print("\nTesting x = 1:")
x = ONE
x_squared = multiply(x, x)
print(f"  1² = {x_squared}")
print(f"  1² = 1? {x_squared == x}")
if x_squared == x:
    roots.append('1')
    steps.append("x = 1 is a root: 1² = 1")

print("\nTesting x = I:")
x = I
x_squared = multiply(x, x)
print(f"  I² = {x_squared}")
print(f"  I² = I? {x_squared == x}")
if x_squared == x:
    roots.append('I')
    steps.append("x = I is a root: I² = I (by P11)")

print(f"\nTotal roots found: {len(roots)}")
print(f"Roots: {', '.join(roots)}")
steps.append(f"Total: {len(roots)} distinct roots")

assert len(roots) > 2, "Should have more than 2 roots!"
print("\n∴ Polynomial has more roots than its degree! ✓")

add_counterexample(
    'CE7',
    'Polynomial Has More Roots Than Degree',
    'Field Axiom: Degree bounds number of roots',
    'x² = x has three solutions: 0, 1, and I',
    steps,
    'Polynomial equations can have more roots than their degree'
)

print("\n" + "="*50)
```

## CE8: Subtraction Paradox (Resolution)

```python
print("=== CE8: Subtraction Paradox ===")
print("\nApparent Paradox: I + I = I but I - I = 0")
print("\nResolution:")

steps = []

# Show I + I = I
result1 = add(I, I)
print(f"\n1. I + I = {result1} (by P12)")
steps.append(f"I + I = {result1} (by P12)")

# Show I - I = 0
result2 = add(I, NEG_I)
print(f"2. I + (-I) = {result2} (by P5, additive inverse)")
steps.append(f"I + (-I) = {result2} (by P5)")

print("\n3. These are not contradictory because:")
print("   - In I + I, both I's are the same element added to itself")
print("   - In I + (-I), -I is the INVERSE of I, not another copy of I")
steps.append("(-I) is the additive inverse, not a second copy of I")

print("\n4. The axiom P5 (additive inverse) takes precedence")
print("   over the idempotency property P12.")
steps.append("P5 (inverse) takes precedence over P12 (idempotency)")

print("\n∴ No contradiction; careful distinction required. ✓")

add_counterexample(
    'CE8',
    'Subtraction Paradox (Resolved)',
    'N/A - This demonstrates careful axiom interaction',
    'I + I = I but I - I = 0 (not a contradiction)',
    steps,
    'Additive inverse is distinct from idempotency'
)

print("\n" + "="*50)
```

## CE9: Exponentiation Doesn't Grow

```python
print("=== CE9: Exponentiation Doesn't Grow ===")
print("\nField Property: For a > 1: aⁿ grows with n")
print("\nCounterexample with I:")

steps = []

# Compute powers of I
powers = {}
current = I

for n in range(1, 6):
    if n == 1:
        current = I
    else:
        current = multiply(current, I)
    
    powers[n] = current
    print(f"I^{n} = {current}")
    steps.append(f"I^{n} = {current}")

# Verify all are equal
all_equal = all(p == I for p in powers.values())
print(f"\nAll powers equal I: {all_equal}")
assert all_equal, "Powers should all be I!"
steps.append("All powers equal I (by repeated application of P11)")

print("\n∴ I is a fixed point under exponentiation. ✓")

add_counterexample(
    'CE9',
    'Exponentiation Does Not Grow',
    'Field Property: Powers grow with exponent',
    'All positive powers of I equal I',
    steps,
    'I is a fixed point under exponentiation'
)

print("\n" + "="*50)
```

## Export Counterexamples

```python
# Export to JSONL (JSON Lines format)
output_path = '../data/examples/counterexamples.jsonl'
os.makedirs(os.path.dirname(output_path), exist_ok=True)

with open(output_path, 'w') as f:
    for ce in counterexamples:
        f.write(json.dumps(ce) + '\n')

print(f"\n✓ Exported {len(counterexamples)} counterexamples to:")
print(f"  {output_path}")

# Also create a summary
summary = {
    'total_counterexamples': len(counterexamples),
    'counterexample_ids': [ce['id'] for ce in counterexamples],
    'axioms_tested': list(set(ce['axiom_violated'] for ce in counterexamples))
}

summary_path = '../data/examples/counterexamples_summary.json'
with open(summary_path, 'w') as f:
    json.dump(summary, f, indent=2)

print(f"✓ Summary exported to: {summary_path}")
```
